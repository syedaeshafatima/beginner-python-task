```python
#TEMPERATURE CONVERTER
celsius=int(input('Enter temperature in Celsius'))
fahrenheit = (celsius * 9/5) + 32
if fahrenheit <10:
  fahrenheit ='Very cold'
elif fahrenheit >= 10 and fahrenheit <=19:
  fahrenheit = 'Cold'
elif fahrenheit >=20 and fahrenheit <= 29:
  fahrenheit = 'Normal'
else:
  fahrenheit >= 30
  fahrenheit = 'Hot'
print(fahrenheit, '°C')

```

```python
#Grocery Shopping Bill
totalproducts= int(input('Enter total products'))
totalbill = 0
for i in range(totalproducts):
  name = input('Enter product name')
  price = int(input(f'Enter the price of {name}:'))
  quantity = int(input(f'Enter product quantity {name}:'))
  subtotal = (price*quantity)
  totalbill = totalbill+subtotal
  if totalbill>=5000:
    discount=(totalbill*10/100)
  else:
    discount=0
  print('total amount:', totalbill)


```

```python
#Employee Weekly Salary
name=input('Enter your name')
hours = float(input('Enter total working hours'))
rate = float(input('Enter hourly rate'))
if hours<= 40:
  salary=hours*rate
  overtimesalary=0
else:
  salary = 40 * rate
  overtimehours = hours - 40
  overtimesalary = overtimehours * rate * 1.5
totalsalary= salary+overtimesalary

print("Employee Name:", name)
print("Salary:", salary)
print("Overtime Salary:", overtimesalary)
print("Total Weekly Salary:", totalsalary)
```
